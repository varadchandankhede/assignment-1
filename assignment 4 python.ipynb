```python
m1 = []
m2 = []
m3 = []
# Input  m1
print("Enter elements for m1:")
for i in range(2):
    row = list(map(int, input().split()))
    m1.append(row)

# Input  m2
print("Enter elements for m2:")
for i in range(2):
    row = list(map(int, input().split()))
    m2.append(row)

# Input m3
print("Enter elements for m3:")
for i in range(2):
    row = list(map(int, input().split()))
    m3.append(row)

print("m1 =", m1)
print("m2 =", m2)
print("m3 =", m3)

# Matrix addition
result = []

for i in range(2):
    row = []
    for j in range(len(m1[i])):
        row.append(m1[i][j] + m2[i][j] + m3[i][j])
    result.append(row)

print("THE ADDITION OF MATRICES IS =", result)
```
