```python
days_in_month = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

d1 = int(input("Enter day of the first day: "))
m1 = int(input("Enter month of the first month: "))
y1 = int(input("Enter year of the first year: "))

d2 = int(input("\nEnter day of the second day: "))
m2 = int(input("Enter month of the second month: "))
y2 = int(input("Enter year of the second year: "))

if (y1, m1, d1) > (y2, m2, d2):
    from_date = (d2, m2, y2)
    to_date = (d1, m1, y1)
else:
    from_date = (d1, m1, y1)
    to_date = (d2, m2, y2)

d1, m1, y1 = from_date
d2, m2, y2 = to_date

days = 0

for y in range(y1, y2):
    if (y % 4 == 0 and y % 100 != 0) or (y % 400 == 0):
        days += 366
    else:
        days += 365

for m in range(1, m1):
    days -= days_in_month[m - 1]
    if m == 2 and ((y1 % 4 == 0 and y1 % 100 != 0) or (y1 % 400 == 0)):
        days -= 1

days -= d1

for m in range(1, m2):
    days += days_in_month[m - 1]
    if m == 2 and ((y2 % 4 == 0 and y2 % 100 != 0) or (y2 % 400 == 0)):
        days += 1

days += d2

print("\nNumber of days between the dates:", days)
```
